## Cargar EXCEL con Pandas

Pandas es una biblioteca de Python para trabajar con datos tabulares. Permite cargar archivos de Excel, organizar la información en DataFrames y analizarla fácilmente.

```python
# Dependencias para cargar y explorar los datos.
# %pip install pandas numpy openpyxl
```

```python
import pandas as pd
import numpy as np

# read_excel carga la tabla; head muestra sus primeras cinco filas.
df = pd.read_excel('data/dataset_ventas.xlsx')
# np.mean calcula el promedio de la columna cantidad.
print(f'Cantidad promedio: {np.mean(df["cantidad"]):.2f}')
df.head()
```

## Manejo de Valores Faltantes y Duplicados

```python
# 1. Identificar valores nulos en el dataset
df.isnull().sum()
```

```python
# 2. Eliminar valores nulos en el dataset
df.dropna()
# dropna devuelve una tabla sin filas incompletas; df no se modifica.
```

```python
# 3. Rellenar valores nulos en el dataset
# fillna reemplaza los vacíos; aquí se muestra el uso con cero sin modificar df.
# El reemplazo por cero debe tener sentido para la columna que se analiza.
df.fillna(0).head()
```

```python
# 4. Rellenar valores nulos con la media de la columna
df['precio_unitario'] = df['precio_unitario'].fillna(df['precio_unitario'].mean())
# mean calcula la media y fillna reemplaza únicamente los precios ausentes.
```

```python
# 5. Verificar si hay valores duplicados en el dataset
# duplicated crea una máscara; keep=False incluye todas las apariciones.
# Repetir un valor de una columna no implica que la fila completa sea duplicada.
mascara = df.duplicated(subset=['cantidad'], keep=False)
# loc selecciona las filas; DataFrame conserva el resultado como una tabla.
duplicados = pd.DataFrame(df.loc[mascara])
duplicados.sort_values(by='cantidad').head(10)
```
